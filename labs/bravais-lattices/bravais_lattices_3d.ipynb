### Why not a real one-atom and two-atom example for every type?
It would be convenient to show a real material with a one-atom basis and another with a two-atom basis for every Bravais type. **Real materials do not provide such a pair for all 14 types**, and this collection does not relabel more complicated crystals to make them fit.

Instead, it provides **one- and two-atom teaching models for every type**, plus **two real-material cases for every type with their actual basis sizes**: 28 principal cases and two bonus semiconductor structures, zincblende GaAs and wurtzite GaN. Four of the selected real-material pairs have exactly one and two basis atoms; the remaining pairs are explicitly labeled otherwise. A few of the structures require high pressure or cryogenic conditions and are not semiconductor-device phases.

**Basis size here always means the number of atoms in a primitive cell of the complete, species-resolved crystal—not the number of chemical elements, and not the number of spheres drawn in a conventional cell.**

## 1 · What is being repeated?
For primitive vectors $\mathbf a_1,\mathbf a_2,\mathbf a_3$,
$$
\mathbf R=n_1\mathbf a_1+n_2\mathbf a_2+n_3\mathbf a_3,\qquad n_i\in\mathbb Z,
$$
and a basis of $N_b$ atoms at $\mathbf p_\alpha$ gives
$$
\boxed{\mathbf r_{n_1n_2n_3,\alpha}=\mathbf R+\mathbf p_\alpha.}
$$
The primitive cell contains **one Bravais lattice point**, but it can contain many atoms. For the settings used here,
$$
N_\text{atoms, conventional}=N_\text{lattice points, conventional}\;N_b.
$$

| Centering | Lattice points per conventional cell | Extra translations in conventional fractional coordinates |
|---|---:|---|
| P | 1 | None |
| C | 2 | $(1/2,1/2,0)$ |
| I | 2 | $(1/2,1/2,1/2)$ |
| F | 4 | $(0,1/2,1/2),(1/2,0,1/2),(1/2,1/2,0)$ |
| R, hexagonal axes | 3 | $(2/3,1/3,1/3),(1/3,2/3,2/3)$ |

For `hR`, the **primitive cell is rhombohedral**; its convenient hexagonal conventional cell has three lattice points. For `hP`, the viewer uses the primitive 120° parallelepiped, not a larger six-sided prism.

The gallery is organized by **lattice systems**. Lattice type describes translations; crystal point symmetry is a different level of description. A hypothetical decoration is not being assigned the space group of a real material.

References: [IUCr Bravais-lattice nomenclature](https://www.iucr.org/who-we-are/commissions/commission-on-crystallographic-nomenclature/published-reports/bravais); [AFLOW Bravais-lattice tutorial](https://aflow.org/p/Tutorials/Crystallography_Part3_Bravis/).

```python
"""Small geometry helpers. All Cartesian cell vectors are ROWS of a 3 x 3 matrix."""
from itertools import product
import math
import numpy as np


def cell_from_parameters(parameters):
    """Convert (a,b,c,alpha,beta,gamma) to a right-handed row-vector cell.

    Lengths retain their input units. Angles are in degrees; alpha is the
    b-c angle, beta the a-c angle, and gamma the a-b angle.
    """
    p = np.asarray(parameters, dtype=float)
    if p.shape != (6,) or not np.isfinite(p).all() or np.any(p[:3] <= 0):
        raise ValueError('Provide six finite parameters, with positive lengths.')
    a, b, c = p[:3]
    alpha, beta, gamma = np.deg2rad(p[3:])
    if not np.all((p[3:] > 0) & (p[3:] < 180)) or abs(np.sin(gamma)) < 1e-10:
        raise ValueError('Cell angles must lie strictly between 0 and 180 degrees.')
    cx = c * np.cos(beta)
    cy = c * (np.cos(alpha) - np.cos(beta) * np.cos(gamma)) / np.sin(gamma)
    cz2 = c*c - cx*cx - cy*cy
    if cz2 <= 0:
        raise ValueError('These parameters do not define a positive-volume 3D cell.')
    return np.array([[a, 0., 0.],
                     [b*np.cos(gamma), b*np.sin(gamma), 0.],
                     [cx, cy, np.sqrt(cz2)]])


def wrap_fractional(f, tolerance=1e-7):
    f = np.mod(np.asarray(f, dtype=float), 1.0)
    f[np.isclose(f, 1.0, atol=tolerance, rtol=0)] = 0.0
    f[np.isclose(f, 0.0, atol=tolerance, rtol=0)] = 0.0
    return f


def periodic_unique(sites):
    """Deduplicate sites modulo lattice translations, preserving species."""
    result = {}
    for site in sites:
        f = wrap_fractional(site['f'])
        key = (site['element'], *np.round(f, 7))
        result[key] = {'element': site['element'], 'f': f.tolist()}
    return list(result.values())


def primitive_basis(material, lattice):
    """Recover the primitive basis from the COMPLETE conventional-cell sites.

    C: conventional row-vector matrix. P = T @ C: primitive matrix.
    The same Cartesian point satisfies f_conventional @ C = f_primitive @ P,
    so f_primitive = f_conventional @ inverse(T).
    Centering duplicates become equivalent and are removed by species.
    """
    T = np.asarray(lattice['transform'])
    invT = np.linalg.inv(T)
    return periodic_unique([
        {'element': s['element'], 'f': np.asarray(s['f']) @ invT}
        for s in material['sites']
    ])


def repeated_crystal(material, cell='primitive', repeats=(2, 2, 2), boundary=False):
    """Return (symbols, Cartesian coordinates) for a repeated crystal.

    With boundary=False the crop is half-open and contains exactly
    product(repeats) times the atom count of the chosen cell.
    boundary=True adds periodic copies on the outer faces when needed.
    The result is a finite visualization crop, not a physical surface model.
    """
    if cell not in ('primitive', 'conventional'):
        raise ValueError("cell must be 'primitive' or 'conventional'.")
    if (len(repeats) != 3 or any(not isinstance(n, (int, np.integer)) or n < 1
                                 or n > 20 for n in repeats)):
        raise ValueError('repeats must contain three integers between 1 and 20.')
    matrix = np.asarray(material[cell + '_matrix'], dtype=float)
    sites = material['primitive_sites'] if cell == 'primitive' else material['sites']
    if math.prod(repeats) * len(sites) > 200000:
        raise ValueError('Requested crop is too large for this teaching function.')
    symbols, positions = [], []
    for site in sites:
        f = np.asarray(site['f'])
        translations = [range(n + int(boundary and abs(f[i]) < 1e-7))
                        for i, n in enumerate(repeats)]
        for n in product(*translations):
            # THIS IS THE CENTRAL "LATTICE + BASIS = CRYSTAL" OPERATION:
            # r = (n1,n2,n3) @ matrix + (u,v,w) @ matrix = R + p_alpha.
            r = (np.asarray(n, dtype=float) + f) @ matrix
            symbols.append(site['element'])
            positions.append(r)
    return symbols, np.asarray(positions)


def validate_collection(data):
    """Geometric consistency checks, not a replacement for source verification."""
    by_lattice = {l['code']: l for l in data['lattices']}
    assert len(by_lattice) == 14
    for lattice in data['lattices']:
        T = np.asarray(lattice['transform'])
        assert np.isclose(abs(np.linalg.det(T)), 1 / len(lattice['centers']))
        assert len(lattice['materials']) == 2
    for material in data['materials']:
        l = by_lattice[material['code']]
        C = cell_from_parameters(material['parameters'])
        P = np.asarray(l['transform']) @ C
        assert np.allclose(C, material['conventional_matrix'])
        assert np.allclose(P, material['primitive_matrix'])
        assert material['n_conventional'] == len(material['sites'])
        assert material['n_basis'] * len(l['centers']) == len(material['sites'])
        recovered = primitive_basis(material, l)
        assert len(recovered) == material['n_basis']
        key = lambda s: (s['element'], *np.round(wrap_fractional(s['f']), 7))
        assert {key(s) for s in recovered} == {key(s) for s in material['primitive_sites']}
        conventional_keys = {key(s) for s in material['sites']}
        for t in l['centers']:
            shifted = [{'element': s['element'], 'f': np.asarray(s['f']) + t}
                       for s in material['sites']]
            assert {key(s) for s in shifted} == conventional_keys
        # Verify that the supplied basis has no missing pure translation.
        # Any extra translation would map the first atom to another of the
        # same species. Test every such candidate in primitive coordinates.
        first = recovered[0]
        primitive_keys = {key(s) for s in recovered}
        for target in recovered[1:]:
            if target['element'] != first['element']:
                continue
            t = np.asarray(target['f']) - np.asarray(first['f'])
            shifted = [{'element': s['element'], 'f': np.asarray(s['f']) + t}
                       for s in recovered]
            if {key(s) for s in shifted} == primitive_keys:
                raise AssertionError(material['id'] + ': nonprimitive basis detected')
        symbols, xyz = repeated_crystal(material, 'primitive', (2, 2, 2))
        assert len(symbols) == 8 * material['n_basis'] and np.isfinite(xyz).all()
    return f"Passed: 14 lattice types; {len(data['materials'])} material models; cell volumes, basis counts, centering, minimal translations and repetition."
```

## 4 · GIF fallback: rotate all 14 point lattices together
This gallery contains **lattice points only**, not the 28 real materials. It is useful in notebook viewers that cannot execute JavaScript and can be inserted into lecture slides. The `hR` tile uses hexagonal conventional axes.

The next function is a small reproducible renderer using NumPy and Pillow. It writes both a rotating GIF and a still PNG. No FFmpeg is required.

```python
from pathlib import Path
import math
from itertools import product
import numpy as np
from PIL import Image as PILImage, ImageDraw, ImageFont


def make_lattice_atlas_gif(data, output, frames=48, duration_ms=90):
    """Save a rotating gallery of the 14 POINT lattices (not material atoms).

    Uses orthographic projection and Pillow only. The main browser viewer
    provides arbitrary dragging and all real-material structures.
    """
    if frames < 8 or frames > 180:
        raise ValueError('Use 8 to 180 frames for this small teaching animation.')
    output = Path(output)
    def font(size, bold=False):
        try:
            return ImageFont.truetype('DejaVuSans-Bold.ttf' if bold else 'DejaVuSans.ttf', size)
        except OSError:
            return ImageFont.load_default()
    title_font, code_font, small_font = font(25,True), font(20,True), font(13)
    corners = np.array(list(product([0,1], repeat=3)), dtype=float)
    edges = [(i,j) for i in range(8) for j in range(i+1,8)
             if np.count_nonzero(corners[i] != corners[j]) == 1]
    width, height = 1400, 640
    models=[]
    for lattice in data['lattices']:
        M=np.asarray(lattice['matrix']);points=[]
        for f in lattice['centers']:
            axes=[[v] + ([1.0] if abs(v)<1e-8 else []) for v in f]
            points.extend(np.array(q) @ M for q in product(*axes))
        verts=corners@M;center=verts.mean(axis=0)
        extent=np.linalg.norm(verts-center,axis=1).max()
        models.append((lattice,verts-center,np.asarray(points)-center,extent))
    images=[]
    for iframe in range(frames):
        im=PILImage.new('RGB',(width,height),'#f2f6f8');d=ImageDraw.Draw(im)
        d.rectangle([0,0,width,83],fill='#153149')
        d.text((24,17),'14 BRAVAIS LATTICES IN 3D',font=title_font,fill='white')
        d.text((25,52),'Professor Nano Labs | Lattice points only: every point has the same translational environment.',font=small_font,fill='#c7e2eb')
        yaw=2*math.pi*iframe/frames+.6;pitch=.4
        ca,sa,cp,sp=math.cos(yaw),math.sin(yaw),math.cos(pitch),math.sin(pitch)
        def project(x):
            X=ca*x[:,0]-sa*x[:,1];q=sa*x[:,0]+ca*x[:,1]
            return np.column_stack([X,cp*x[:,2]+sp*q,sp*x[:,2]-cp*q])
        for index,(lattice,verts,pts,extent) in enumerate(models):
            col,row=index%7,index//7;x0=15+col*198;y0=99+row*244
            d.rounded_rectangle([x0,y0,x0+183,y0+226],radius=9,fill='white',outline='#d5e2e9')
            d.text((x0+12,y0+10),lattice['code'],font=code_font,fill='#24619c')
            label=lattice['system']
            d.text((x0+12,y0+38),label,font=small_font,fill='#48677e')
            v=project(verts);a=project(pts);sc=61/extent
            xy=lambda p:(x0+91+p[0]*sc,y0+128-p[1]*sc)
            for i,j in edges:d.line([xy(v[i]),xy(v[j])],fill='#899da9',width=2)
            for p in sorted(a,key=lambda q:q[2]):
                x,y=xy(p);r=4.3
                d.ellipse([x-r,y-r,x+r,y+r],fill='#2671b0',outline='#184b78')
                d.ellipse([x-2,y-3,x+1,y],fill='#a8d4eb')
            ctype={'P':'primitive','C':'base-centered','I':'body-centered','F':'face-centered','R':'hexagonal axes'}[lattice['centering']]
            d.text((x0+12,y0+192),ctype,font=small_font,fill='#48677e')
            d.text((x0+12,y0+208),f"{len(lattice['centers'])} point(s) / conv. cell",font=small_font,fill='#48677e')
        d.text((24,602),'Centering points belong to the lattice. They are not additional basis atoms. All 14 have one lattice point per primitive cell.',font=small_font,fill='#48677e')
        images.append(im)
    output.parent.mkdir(parents=True,exist_ok=True)
    images[0].save(output,save_all=True,append_images=images[1:],duration=duration_ms,loop=0,optimize=False)
    images[0].save(output.with_suffix('.png'))
    return output
```

### Five useful classroom comparisons
**Cu → Si → GaAs:** keep the face-centered cubic translation lattice, but compare a one-atom basis with two identical atoms and then two different atoms. Use the material cards' atom counts rather than counting boundary spheres.

**W → CsI:** W has a body-centered cubic translation lattice with a one-atom basis. CsI has a primitive cubic lattice with two different atoms. A translation must preserve species, not merely occupied positions.

**Ru → GaN:** HCP Ru has two basis atoms; wurtzite GaN has four. Both use a primitive hexagonal translation lattice. Two chemical elements in GaN do not imply two basis atoms.

**Ga → black phosphorus:** both selected elemental examples have four atoms per primitive basis. “One element” and “one-atom basis” are different statements.

**GaAs → high-pressure GaAs:** compare the ordinary zincblende semiconductor structure with the historically reported high-pressure orthorhombic model. A chemical formula alone does not specify a phase or justify transferring device applications between phases.

## Model limitations and responsible interpretation
All atoms are fixed at idealized, full-occupancy positions. Thermal motion, defects, dopants, surfaces, bonding forces, and electronic bands are not calculated. Several structures use representative rounded parameters; the cited original studies are needed for quantitative materials calculations.

Material relevance is deliberately separated into **semiconductors, insulating device components, metallic contacts/interconnects, synthesis precursors, and structural references**. A precursor's crystal structure is not the structure of the device made from it. A bulk parent structure is not a thin-film device refinement. High-pressure GaAs is not a substitute for the ordinary device phase.

The 14-lattice inventory and the 30 stored structural models pass the geometric checks in this notebook. Those checks test internal consistency; they are not an independent experimental validation of a crystallographic phase assignment.

**Key takeaway:** the Bravais lattice specifies the translations; the basis tells us what is repeated.
